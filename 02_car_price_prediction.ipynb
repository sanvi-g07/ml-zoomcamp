# Train Linear Regression Model

```python
def train_linear_regression(X, y):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])

    XTX = X.T.dot(X)
    XTX_inv = np.linalg.inv(XTX)
    w = XTX_inv.dot(X.T).dot(y)

    return w[0], w[1:]
```

#RMSE

```python
def rmse(y, y_pred):
    error = y_pred - y
    mse = (error ** 2).mean()
    return np.sqrt(mse)
```

```python
categorical_vars = ['make', 'engine_fuel_type', 'transmission_type', 'driven_wheels', 'market_category', 'vehicle_size', 'vehicle_style']
```

```python
categories = {}

for c in categorical_vars:
  categories[c] = list(df[c].value_counts().head().index)
```

```python
categories
```

```python
def prepare_X(df):
  df = df.copy()

  df['age'] = 2017 - df.year
  features = base + ['age'] #not using append here to ensure base remains unchanged

  for v in [2, 3, 4]:
    df['num_doors_%s' % v] = (df.number_of_doors == v).astype('int')
    features.append('num_doors_%s' % v)

  for c, values in categories.items():
    for v in values:
      df['%s_%s' % (c, v)] = (df[c] == v).astype('int')
      features.append('%s_%s' % (c, v))


  df_num = df[features]
  df_num = df_num.fillna(0)
  X = df_num.values

  return X
```

```python
X_train = prepare_X(df_train)
w0, w = train_linear_regression(X_train, y_train)

X_val = prepare_X(df_val)
y_pred = w0 + X_val.dot(w)

rmse(y_val, y_pred)
```

#Regularization

```python
def train_linear_regression_reg(X, y, r=0.001):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])

    XTX = X.T.dot(X)
    reg = r * np.eye(XTX.shape[0])
    XTX = XTX + reg

    XTX_inv = np.linalg.inv(XTX)
    w = XTX_inv.dot(X.T).dot(y)

    return w[0], w[1:]
```

```python
X_train = prepare_X(df_train)
w0, w = train_linear_regression_reg(X_train, y_train)

X_val = prepare_X(df_val)
y_pred = w0 + X_val.dot(w)

rmse(y_val, y_pred)
```

# Tuning the Model

```python
for r in [0.0, 0.0001, 0.0001, 0.001, 0.01, 0.1, 1, 10]:
  X_train = prepare_X(df_train)
  w0, w = train_linear_regression_reg(X_train, y_train, r=r)

  X_val = prepare_X(df_val)
  y_pred = w0 + X_val.dot(w)

  score = rmse(y_val, y_pred)
  print(r, w0, score)
```
